```python
import os
import subprocess

# --- 1. REEMPLAZA ESTO CON LA RUTA DE TU CARPETA 'val' DEL DRIVE ---
ruta_drive_val = "/content/drive/MyDrive/LipFD/val"


ruta_real = os.path.join(ruta_drive_val, "0_real")

if not os.path.exists(ruta_real):
    print(f"❌ ERROR: Colab no encuentra la carpeta en: '{ruta_real}'")
    print("Revisa si te faltó poner 'MyDrive/' o si hay algún espacio en los nombres.")
else:
    archivos = [f for f in os.listdir(ruta_real) if os.path.isfile(os.path.join(ruta_real, f))]
    print(f"✅ ¡Carpeta encontrada en Drive! Contiene {len(archivos)} archivos.")
    print(f"Muestra de los primeros 3 archivos: {archivos[:3]}")

    exts = set(f.split('.')[-1].lower() for f in archivos if '.' in f)

    if 'png' in exts or 'jpg' in exts or 'jpeg' in exts:
        print("\n🎉 ¡Son imágenes! El formato es 100% el que pide el modelo.")
        print("Lanzando la validación directamente contra tu Google Drive...")

        cmd = f"/content/venv_lipfd/bin/python /content/LipFD/validate.py --real_list_path '{ruta_drive_val}/0_real' --fake_list_path '{ruta_drive_val}/1_fake' --ckpt /content/LipFD/checkpoints/ckpt.pth"
        subprocess.run(cmd, shell=True)

    elif 'mp4' in exts:
        print("\n⚠️ DETENIDO: Tus datos en el Drive son videos (.mp4).")
        print("Como dicta utils.py, el modelo LipFD ignora los .mp4. Necesitas convertirlos a espectrogramas .png usando preprocess.py antes de poder validar.")
    else:
        print(f"\n❓ Formato desconocido. Las extensiones detectadas fueron: {exts}")
```

```python
# Abre el archivo de validación
with open('/content/LipFD/validate.py', 'r') as f:
    codigo = f.read()

# Inyectamos 'model.to(device)' obligatoriamente dentro de la función validate()
linea_original = 'device = torch.device(f"cuda:{gpu_id[0]}" if torch.cuda.is_available() else "cpu")'
linea_parcheada = linea_original + '\n    model.to(device)'

codigo_corregido = codigo.replace(linea_original, linea_parcheada)

with open('/content/LipFD/validate.py', 'w') as f:
    f.write(codigo_corregido)

print("¡Parche aplicado! El modelo ahora se moverá a la GPU obligatoriamente.")
```

```python
import torch

# 1. Pega aquí la ruta exacta de tu ckpt.pth guardado en tu Google Drive:
ruta_drive_ckpt = "/content/drive/MyDrive/LipFD/ckpt.pth"

print("Abriendo el archivo de 890 MB desde Google Drive...")
checkpoint_pesado = torch.load(ruta_drive_ckpt, map_location="cpu")

# 2. Extraemos estrictamente los pesos limpios de la IA
pesos_limpios = checkpoint_pesado["model"]

# 3. Lo guardamos en el disco temporal listo para subirlo
nombre_para_hf = "lipfd_pytorch_weights.pt"
torch.save(pesos_limpios, nombre_para_hf)

print(f"✅ ¡Extraído con éxito! El archivo '{nombre_para_hf}' ya está listo.")
```

```python
import torch
from huggingface_hub import HfApi, login

# --- PASO A: EXTRACCIÓN FÍSICA DEL ARCHIVO ---
ruta_drive_ckpt = "/content/drive/MyDrive/LipFD/ckpt.pth"
nombre_archivo = "lipfd_pytorch_weights.pt"

print("1️⃣ Extrayendo pesos limpios del archivo de 890 MB en Drive...")
checkpoint_pesado = torch.load(ruta_drive_ckpt, map_location="cpu")

torch.save(checkpoint_pesado["model"], nombre_archivo)
print(f"✅ ¡Archivo '{nombre_archivo}' creado físicamente en Colab!")


# --- PASO B: SUBIDA A HUGGING FACE ---
print("\n2️⃣ Autenticando con Hugging Face...")
login()
api = HfApi()

print(f"\n3️⃣ Subiendo archivo hacia el repositorio Alyce12/LipFD...")
api.upload_file(
    path_or_fileobj=nombre_archivo,
    path_in_repo=nombre_archivo,
    repo_id="Alyce12/LipFD",
    repo_type="model"
)

print("\n🚀 ¡VICTORIA ABSOLUTA! Revisa tu repositorio en la web de Hugging Face.")
```
